Convert Timestamp properly

```python
df["cdTimestamp"]=pd.to_datetime(df["cdTimestamp"],errors='coerce')
```

```python
print(df["cdTimestamp"].dtype)
print("Invalid timestamps:",df["cdTimestamp"].isna().sum())
```

```python
print(df["NDVI"].head(20))
print(df["NDVI"].unique()[:20])
```

```python
bad_ndvi = pd.to_numeric(df["NDVI"], errors="coerce").isna()

print("Invalid NDVI values:", bad_ndvi.sum())
print(df.loc[bad_ndvi])
```

```python
df["NDVI"] = df["NDVI"].astype(str).str.replace("git", "", regex=False)
df["NDVI"] = pd.to_numeric(df["NDVI"], errors="coerce")
```

```python
print(df["NDVI"].dtype)
print("missing NDVI:",df["NDVI"].isna().sum())
```
